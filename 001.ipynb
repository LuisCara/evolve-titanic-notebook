```python
# Import necessary libraries
import pandas as pd
import numpy as np

# Read the Titanic dataset
df = pd.read_csv('data/titanic.csv')

# Display the first few rows of the dataset
print("First few rows of the Titanic dataset:")
df.head()
```

## Relación entre variables

```python
# Import visualization libraries
import seaborn as sns
import matplotlib.pyplot as plt

# Set the style for better visualizations
plt.style.use('classic')  # Using a valid matplotlib style
sns.set_theme()  # This is the correct way to set seaborn style

# Create a figure with subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('Titanic Survival Analysis', fontsize=16, y=1.02)

# 1. Passenger Class vs Survival
sns.barplot(data=df, x='Pclass', y='Survived', ax=axes[0,0])
axes[0,0].set_title('Survival Rate by Passenger Class')
axes[0,0].set_xlabel('Passenger Class')
axes[0,0].set_ylabel('Survival Rate')

# 2. Age vs Survival
sns.boxplot(data=df, x='Survived', y='Age', ax=axes[0,1])
axes[0,1].set_title('Age Distribution by Survival Status')
axes[0,1].set_xlabel('Survived (0=No, 1=Yes)')
axes[0,1].set_ylabel('Age')

# 3. Number of Siblings/Spouses vs Survival
sns.barplot(data=df, x='SibSp', y='Survived', ax=axes[1,0])
axes[1,0].set_title('Survival Rate by Number of Siblings/Spouses')
axes[1,0].set_xlabel('Number of Siblings/Spouses')
axes[1,0].set_ylabel('Survival Rate')

# 4. Fare vs Survival
sns.boxplot(data=df, x='Survived', y='Fare', ax=axes[1,1])
axes[1,1].set_title('Fare Distribution by Survival Status')
axes[1,1].set_xlabel('Survived (0=No, 1=Yes)')
axes[1,1].set_ylabel('Fare')

# Adjust layout
plt.tight_layout()
plt.show()

# Additional visualization for Age distribution
plt.figure(figsize=(12, 6))
sns.kdeplot(data=df[df['Survived']==0]['Age'], label='Not Survived', fill=True)
sns.kdeplot(data=df[df['Survived']==1]['Age'], label='Survived', fill=True)
plt.title('Age Distribution by Survival Status')
plt.xlabel('Age')
plt.ylabel('Density')
plt.legend()
plt.show()
```
